# Numbers and Mathematical Operations This notebook explores numerical data types, arithmetic operations, type conversion, and augmented assignment operators.

## 1. Integers & Floating Point Numbers Python distinguishes between whole numbers (int) and decimals (float). We can also convert between types and round values.

```python

#integer examples
my_int_1 = 56
my_int_2 = -4

print(type(my_int_1)) # <class 'int'>
print(type(my_int_2)) # <class 'int'>

#floating point examples
my_float_1 = -12.0
my_float_2 = 4.9

print(type(my_float_1)) # <class 'float'>
print(type(my_float_2)) # <class 'float'>

#Type Conversion between Integers and Floating Point Numbers

# integer to float
my_int = 10
converted_float = float(my_int)
print('Converted Integer to Float:', converted_float) # Converted Integer to Float: 10.0
print('Type:', type(converted_float)) # Type: <class 'float'>  

# float to integer
my_float = 9.8
converted_int = int(my_float)
print('Converted Float to Integer:', converted_int) # Converted Float to Integer: 9
print('Type:', type(converted_int)) # Type: <class 'int'>

#Rounding Floating Point Numbers
my_float = 5.6789

rounded_float_2 = round(my_float, 2)
print('Rounded to 2 decimal places:', rounded_float_2) # Rounded to 2 decimal places: 5.68

rounded_float_0 = round(my_float)
print('Rounded to nearest integer:', rounded_float_0) # Rounded to nearest integer: 6   
print('Type:', type(rounded_float_0)) # Type: <class 'float'>

#Absolute Value
num = -15

absolute_value = abs(num)
print(absolute_value) # 15

```

## 2. Arithmetic Operators Beyond standard math, Python provides Floor Division (//), Modulus (%), and Exponentiation (**).

```python

#standard arithmetic with integers

my_int_1 = 56
my_int_2 = 12

# addition
sum_ints = my_int_1 + my_int_2
print('Integer Addition:', sum_ints) # Integer Addition: 68

# Subtraction
diff_ints = my_int_1 - my_int_2
print('Integer Subtraction:', diff_ints) # Integer Subtraction: 44

# Multiplication
prod_ints = my_int_1 * my_int_2
print('Integer Multiplication:', prod_ints) # Integer Multiplication: 672

# Division
quot_ints = my_int_1 / my_int_2
print('Integer Division:', quot_ints) # Integer Division: 4.666666666666667


#arithmetic with floating point numbers

my_float_1 = 5.4
my_float_2 = 12.0

# addition
float_addition = my_float_1 + my_float_2
print('Float Addition:', float_addition) # Float Addition: 17.4

# Subtraction
float_subtraction = my_float_1 - my_float_2
print('Float Subtraction:', float_subtraction) # Float Subtraction: -6.6

# Multiplication
float_multiplication = my_float_1 * my_float_2
print('Float Multiplication:', float_multiplication) # Float Multiplication: 64.8

# Division
float_division = my_float_1 / my_float_2
print('Float Division:', float_division) # Float Division: 0.45 


#Mixed arithmetic with integers and floating point numbers

my_int = 7
my_float = 3.5 

# addition
mixed_addition = my_int + my_float
print('Mixed Addition:', mixed_addition) # Mixed Addition: 10.5

# Subtraction
mixed_subtraction = my_int - my_float
print('Mixed Subtraction:', mixed_subtraction) # Mixed Subtraction: 3.5

# Multiplication
mixed_multiplication = my_int * my_float
print('Mixed Multiplication:', mixed_multiplication) # Mixed Multiplication: 24.5

# Division
mixed_division = my_int / my_float
print('Mixed Division:', mixed_division) # Mixed Division: 2.0

#Modulus Operator
my_int_1 = 56
my_int_2 = 12

my_float_1 = 5.4
my_float_2 = 12.0

mod_ints = my_int_1 % my_int_2
mod_floats = my_float_2 % my_float_1

print('Integer Modulus:', mod_ints) # Integer Modulus: 8
print('Float Modulus:', mod_floats) # Float Modulus: 1.1999999999999993

#  // (Floor Division)
my_int_1 = 56
my_int_2 = 12

my_float_1 = 5.4
my_float_2 = 12.0

floor_div_ints = my_int_1 // my_int_2
floor_div_floats = my_float_2 // my_float_1

print('Integer Floor Division:', floor_div_ints) # Integer Floor Division: 4
print('Float Floor Division:', floor_div_floats) # Float Floor Division: 2.0

#** (Exponentiation)
my_int_1 = 56
my_int_2 = 12

my_float_1 = 5.4
my_float_2 = 12.0

exp_ints = my_int_1 ** my_int_2
exp_floats = my_float_1 ** my_float_2

print('Integer Exponentiation:', exp_ints) # Integer Exponentiation: 951166013805414055936
print('Float Exponentiation:',  exp_floats) # Float Exponentiation: 614787626.1765089

#pow() Function
result_1 = pow(2, 3)  # Equivalent to 2 ** 3
print(result_1)  # 8

result_2 = pow(2, 3, 5)  # (2 ** 3) % 5
print(result_2)  # 3
```

## 3. Augmented Assignment Operators These operators provide a shorthand for updating a variable's value based on its current state.

```python
#Augmented Assignment Operators

# += (Addition Assignment)
my_var = 10
my_var += 5

print(my_var) # 15

# -= (Subtraction Assignment)
my_var = 10
my_var -= 3

print(my_var) # 7

# *= (Multiplication Assignment)
my_var = 10
my_var *= 2

print(my_var) # 20

# /= (Division Assignment)
my_var = 10
my_var /= 2

print(my_var) # 5.0

# %= (Modulus Assignment)
my_var = 10
my_var %= 3
print(my_var) # 1

# //= (Floor Division Assignment)
my_var = 10
my_var //= 3
print(my_var) # 3

# **= (Exponentiation Assignment)
my_var = 2
my_var **= 3
print(my_var) # 8

#String Concatenation with +=
greet = 'Hello'
greet += ' World'

print(greet) # Hello World

#String Subtraction with -=
#greet = 'Hello'
#greet -= ' World'

#print(greet) # TypeError: unsupported operand type(s) for -=: 'str' and 'str'

#String Division with /=
#greet = 'Hello'
#greet /= 'World'

#print(greet) # TypeError: unsupported operand type(s) for /=: 'str' and 'str' 

#String Repetition with *=
greet = 'Hello'
greet *= 3

print(greet) # HelloHelloHello
```

## 4. Important Notes

```python
#No Increment/Decrement: Python does not support ++ or --. Use += 1 instead.
my_var = 5

print(+my_var)   # 5
print(++my_var)  # 5
print(+++my_var) # 5
my_var += 1
print(my_var) # 6

#Floating Point Precision: Note that 5.4 % 12.0 might show as 1.1999... due to how computers handle decimals.

#Division Result: Standard division / always returns a float, even if the numbers divide perfectly (e.g., 4 / 2 = 2.0).
```
