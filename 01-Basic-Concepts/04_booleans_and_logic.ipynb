# Booleans, Comparison, and Conditional Logic This notebook covers how Python evaluates truth values and makes decisions based on conditions.

## 1. Comparison Operators These operators compare two values and return a Boolean (TrueorFalse).

```python
# Comparison Operators.

# Greater than, Less than, Equal to, Not equal to, Greater than or equal to, Less than or equal to
print(3 > 4) # False
print(3 < 4) # True
print(3 == 4) # False
print(4 == 4) # True
print(3 != 4) # True
print(3 >= 4) # False
print(3 <= 4) # True

```

## 2. Conditional Statements (if, elif, else) The ifstatement executes a block of code only if a condition is true.elif(else if) allows checking multiple conditions, andelse covers any remaining cases.

```python
#if conditional statements
age = 12

if age >= 18:
    print('You are an adult')
else:
    print('You are not an adult yet') # You are not an adult yet


# if-elif-else conditional statements
age = 12

if age >= 18:
    print('You are an adult')
elif age >= 13:
    print('You are a teenager')
else:
    print('You are a child') # You are a child

# More complex if-elif-else conditional statements

age = 2

if age >= 65:
    print('You are a senior citizen')
elif age >= 30:
    print('You are an adult in your prime')
elif age >= 18:
    print('You are a young adult')
elif age >= 13:
    print('You are a teenager')
elif age >= 3:
    print('You are a young child')
else:
    print('You are a toddler or an infant') # You are a toddler or an infant
```

## 3. Logical Operators Used to combine multiple conditions together.

```python
# Boolean values and logic
print(bool(False)) # False
print(bool(0))  # False
print(bool('')) # False

print(bool(True)) # True
print(bool(1)) # True
print(bool('Hello')) # True

# and operator
is_citizen = True
age = 25
print(is_citizen and age) # 25

is_citizen = True
age = 25

if is_citizen and age >= 18:
    print('You are eligible to vote') # You are eligible to vote
else:
    print('You are not eligible to vote')

# or operator
age = 19
is_employed = False
print(age or is_employed) # 19

age = 19
is_student = True

if age < 18 or is_student:
    print('You are eligible for a student discount') # You are eligible for a student discount
else:
    print('You are not eligible for a student discount')

# not operator
print(not '') # True, because empty string is falsy
print(not 'Hello') # False, because non-empty string is truthy
print(not 0) # True, because 0 is falsy
print(not 1) # False, because 1 is truthy
print(not False) # True, because False is falsy
print(not True) # False, because True is truthy

is_admin = False

if not is_admin:
    print('Access denied for non-administrators.') # Access denied for non-administrators.
else:
    print('Welcome, Administrator!')

```
