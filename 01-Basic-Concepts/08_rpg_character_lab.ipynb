# Lab 08: RPG Character Profile Building a structured character data system using variables and strings

```python
full_dot = '●'
empty_dot = '○'

def create_character(name, strength, intelligence, charisma):
    # 1. التحقق من نوع واسم الشخصية
    if not isinstance(name, str):
        return "The character name should be a string"
    if name == "":
        return "The character should have a name"
    if len(name) > 10:
        return "The character name is too long"
    if " " in name:
        return "The character name should not contain spaces"
    
    # 2. التحقق من نوع البيانات (أرقام صحيحة)
    if type(strength) is not int or type(intelligence) is not int or type(charisma) is not int:
        return "All stats should be integers"
    
    # 3. التحقق من الحدود (1 إلى 4)
    if strength < 1 or intelligence < 1 or charisma < 1:
        return "All stats should be no less than 1"
    if strength > 4 or intelligence > 4 or charisma > 4:
        return "All stats should be no more than 4"
    
    # 4. التحقق من مجموع النقاط
    if (strength + intelligence + charisma) != 7:
        return "The character should start with 7 points"

    # 5. المخرج النهائي المنسق
    line1 = name
    line2 = f"STR {full_dot * strength}{empty_dot * (10 - strength)}"
    line3 = f"INT {full_dot * intelligence}{empty_dot * (10 - intelligence)}"
    line4 = f"CHA {full_dot * charisma}{empty_dot * (10 - charisma)}"
    
    return f"{line1}\n{line2}\n{line3}\n{line4}"

    # تجربة نهائية للتأكد من المخرج المنسق
print(create_character('ren', 4, 2, 1))
```
