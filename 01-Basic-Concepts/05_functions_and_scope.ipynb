# Functions and Scope in Python Functions allow us to write reusable blocks of code. Scope determines where variables can be accessed.

## 1. Defining and Calling Functions A function is defined using the def keyword, followed by a name and parentheses.

```python

name = input('What is your name?') # User types "Kolade" and presses Enter  
print('Hello', name) # Output: Hello Kolade

```

## 2. Parameters and Return Values Functions can take inputs (parameters) and send back a result using the return statement.

```python
#use def function
def hello():
    print('Hello World')
hello() # Hello World

def calculate_sum(a, b):
    print(a + b)
calculate_sum(5, 3) # Output: 8
################################################
#def calculate_sum(a, b):
 #   print(a + b)

#my_sum = calculate_sum(3, 1) # 4
#print(my_sum) # None 
#################################################

#returning values from functions
def calculate_sum(a, b):
    return a + b

my_sum = calculate_sum(3, 1)
print(my_sum) # 4

```

## 3. Global vs. Local Scope

```python
#local scope
def my_func():
    my_var = 10
    print(my_var)
my_func()  # Output: 10
#print(my_var)  # Error: NameError: name 'my_var' is not defined

#global scope
my_var = 20 # Global variable  
def my_func():
    print(my_var)  # Accessing global variable
my_func()  # Output: 20
print(my_var)  # Output: 20


```

## 4. enclosing scope

```python

def outer_func():
    msg = 'Hello there!'
    res = ""  # Declare res in the enclosing scope

    def inner_func():
        nonlocal res  # Allow modification of an enclosing variable
        res = 'How are you?'
        print(msg)  # Accessing msg from outer_func()

    inner_func()
    print(res)  # Now res is accessible and modified

outer_func()

# Output:
# Hello there!
# How are you?
```
