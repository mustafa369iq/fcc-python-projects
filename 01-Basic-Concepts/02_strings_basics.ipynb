```python
# String Immutability

# Single and Double Quotes
my_str_1 = 'Hello'
my_str_2 = "World"

# Multiline Strings
my_str_3 = """Multiline 
string"""
my_str_4 = '''Another 
multiline 
string'''

print ("my_str_1:", my_str_1)
print ("my_str_2:", my_str_2)
print ("my_str_3:", my_str_3)
print ("my_str_4:", my_str_4)

# Handling Quotes in Strings
msg = "It's a sunny day"
quote = 'She said, "Hello World!"'

print("msg:", msg)
print("quote:", quote)

msg1 = 'It\'s a sunny day'
quote1 = "She said, \"Hello!\""

print("msg1:", msg1)
print("quote1:", quote1)

# String Membership Testing
my_str = 'Hello world'

print('Hello' in my_str)  # True
print('hey' in my_str)    # False
print('hi' in my_str)    # False
print('e' in my_str)  # True
print('f' in my_str)  # False

print('Hello' not in my_str)  # False
print('hey' not in my_str)    # True
print('hi' not in my_str)    # True
print('e' not in my_str)  # False
print('f' not in my_str)  # True

# String Length
my_str = 'Hello world'
print(len(my_str))  # 11

# String Indexing
my_str = "Hello world"

print(my_str[0])  # H
print(my_str[6])  # w
print(my_str[-1]) # d
print(my_str[-5]) # o

# String Immutability
greeting = 'hi'
greeting = 'hello'
print(greeting) # hello

greeting = 'hi'
#greeting[0] = 'H' # TypeError: 'str' object does not support item assignment

```
