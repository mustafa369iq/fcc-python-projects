```python
# String Immutability

# Single and Double Quotes
my_str_1 = 'Hello'
my_str_2 = "World"

# Multiline Strings
my_str_3 = """Multiline 
string"""
my_str_4 = '''Another 
multiline 
string'''

print ("my_str_1:", my_str_1)
print ("my_str_2:", my_str_2)
print ("my_str_3:", my_str_3)
print ("my_str_4:", my_str_4)

# Handling Quotes in Strings
msg = "It's a sunny day"
quote = 'She said, "Hello World!"'

print("msg:", msg)
print("quote:", quote)

msg1 = 'It\'s a sunny day'
quote1 = "She said, \"Hello!\""

print("msg1:", msg1)
print("quote1:", quote1)

# String Membership Testing
my_str = 'Hello world'

print('Hello' in my_str)  # True
print('hey' in my_str)    # False
print('hi' in my_str)    # False
print('e' in my_str)  # True
print('f' in my_str)  # False

print('Hello' not in my_str)  # False
print('hey' not in my_str)    # True
print('hi' not in my_str)    # True
print('e' not in my_str)  # False
print('f' not in my_str)  # True

# String Length
my_str = 'Hello world'
print(len(my_str))  # 11

# String Indexing
my_str = "Hello world"

print(my_str[0])  # H
print(my_str[6])  # w
print(my_str[-1]) # d
print(my_str[-5]) # o

# String Immutability
greeting = 'hi'
greeting = 'hello'
print(greeting) # hello

greeting = 'hi'
#greeting[0] = 'H' # TypeError: 'str' object does not support item assignment

```

```python
# Concatenation and Interpolation

# String Concatenation
my_str_1 = 'Hello'
my_str_2 = "World"

str_plus_str = my_str_1 + ' ' + my_str_2
print(str_plus_str) # Hello World

# String and Non-String Concatenation
#name = 'John Doe'
#age = 26

#name_and_age = name + age
#print(name_and_age) # TypeError: can only concatenate str (not "int") to 

#convert non-string to string
name = 'John Doe'
age = 26

name_and_age = name + str(age)
print(name_and_age) # John Doe26

# Using += Operator for Concatenation
name = 'John Doe'
age = 26

name_and_age = name  # Start with the name
name_and_age += str(age)  # Append the age as string

print(name_and_age)  # John Doe26

# String Interpolation with f-Strings
name = 'John Doe'
age = 26
name_and_age = f'My name is {name} and I am {age} years old'
print(name_and_age) # My name is John Doe and I am 26 years old

num1 = 5
num2 = 10
print(f'The sum of {num1} and {num2} is {num1 + num2}') # The sum of 5 and 10 is 15
```

```python
# String Slicing
my_str = "Hello world"

print(my_str[0])  # H
print(my_str[6])  # w
print(my_str[-1]) # d

# Note that the stop index is non-inclusive, so [1:4] just extracted the characters from index 1, and up to, but not including, the character at index 4.
print(my_str[1:4])  # ell
# This extracts everything from index 0 up to (but not including), the character at index 7. And here's what happens if you omit the stop index:
print(my_str[:7])  # Hello w
# This extracts everything from index 3 to the end of the string:
print(my_str[3:])  # lo world
# And this extracts the entire string:
print(my_str[:])  # Hello world
# This extracts every second character from index 0 to index 10 (non-inclusive):
print(my_str[0:11:2])  # Hlowrd
#reverse a string by setting step to -1, and leaving start and stop blank:
print(my_str[::-1]) # dlrow olleH

```

```python
#string Methods

#uppercase.
my_str = "Hello world"
print(my_str.upper())  # HELLO WORLD

uppercase_my_str = my_str.upper()
print(uppercase_my_str)  # HELLO WORLD

#lowercase
print(my_str.lower())  # hello world

lowercase_my_str = my_str.lower()
print(lowercase_my_str)  # hello world

#strip whitespace
print(my_str.strip())  # "hello world"

trimmed_my_str = my_str.strip()
print(trimmed_my_str)  # "hello world"

#replace substring
print(my_str.replace('world', 'there'))  # Hello there

replaced_my_str = my_str.replace('world', 'there')
print(replaced_my_str)  # Hello there

#split string
print(my_str.split(' '))  # ['Hello', 'world'] 

split_my_str = my_str.split(' ')
print(split_my_str)  # ['Hello', 'world']

#Joins strings
words = ['Hello', 'world']
print(' '.join(words))  # Hello world 

joined_str = ' '.join(words)
print(joined_str)  # Hello world

#startswith(prefix) 
print(my_str.startswith('Hello'))  # True  

starts_with_hello = my_str.startswith('Hello')
print(starts_with_hello)  # True

print(my_str.startswith('world'))  # False

starts_with_world = my_str.startswith('world')
print(starts_with_world)  # False  

#endswith(suffix)
print(my_str.endswith('world'))  # True

ends_with_world = my_str.endswith('world')
print(ends_with_world)  # True

print(my_str.endswith('Hello'))  # False

ends_with_hello = my_str.endswith('Hello')
print(ends_with_hello)  # False

#find(substring)
print(my_str.find('world'))  # 6

index_of_world = my_str.find('world')
print(index_of_world)  # 6 


print(my_str.find('there'))  # -1

index_of_there = my_str.find('there')
print(index_of_there)  # -1

#capitalize
print(my_str.capitalize())  # Hello world

capitalized_str = my_str.capitalize()
print(capitalized_str)  # Hello world

#isupper 
print(my_str.isupper())  # False   

is_upper = my_str.isupper()
print(is_upper)  # False

#islower
print(my_str.islower())  # False

is_lower = my_str.islower()
print(is_lower)  # False

#title
print(my_str.title())  # Hello World

titled_str = my_str.title()
print(titled_str)  # Hello World
```
